## 📦 Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("✅ Libraries imported successfully!")
```

## 🐼 Create the Student Dataset

We will create a small table (DataFrame) with information about 12 students:

- **Study_Hours** → hours the student studies per day
- **Attendance** → attendance percentage
- **Homework_Score** → average homework score
- **Previous_Score** → score from the last exam
- **Final_Score** → the most recent final exam score

😉 We will leave a couple of values empty on purpose, so we can practice cleaning the data.

```python
data = {
    "Student": ["Ali", "Sara", "Omar", "Laila", "Yusuf", "Mona",
                "Karim", "Hana", "Adam", "Nour", "Tarek", "Dina"],
    "Study_Hours": [5, 2, 6, 3, 7, 1, 4, 6, 2, 5, 3, 7],
    "Attendance": [95, 70, 98, 80, 100, 60, 85, 92, 75, 90, 65, 97],
    "Homework_Score": [88, 60, 95, 70, 92, 55, 78, 90, 65, 85, 68, 96],
    "Previous_Score": [80, 65, 90, 72, 88, 58, 75, 85, 66, 82, 70, 91],
    "Final_Score": [85, 62, 93, 75, 90, np.nan, 80, 88, 68, 84, np.nan, 94]
}

df = pd.DataFrame(data)

print("✅ Dataset created! It has", df.shape[0], "students.")
```

## 👀 Look at the Data

Let's look at our table using `head()`, and check its size with `shape` and `columns`.

```python
df.head(12)
```

```python
print("Shape (rows, columns):", df.shape)
print("Columns:", df.columns.tolist())
```

We can also use `info()` to see the data types, and `describe()` to get a quick statistical summary.

```python
df.info() 
```

```python
df.describe()
```

## 🧹 Clean the Data

Sometimes our data has missing values. We need to clean them before analyzing the data.

Let's check which values are missing using `isnull().sum()`.

```python
df.isnull()
```

```python
print("Missing values in each column:")

print(df.isnull().sum().sort_values(ascending=False))
```

We can see that `Final_Score` has 2 missing values. Let's fill them with the **average (mean)** Final Score using `fillna()`.

```python
values_to_fill = {
    "Final_Score": df["Final_Score"].mean()
    
  
}
df = df.fillna(value=values_to_fill)

print("Missing values after cleaning:")
print(df.isnull().sum().sort_values(ascending=False))

print("\n All clean! No missing values left.")
```

```python
df.head(12)
```

## 🔢 Simple NumPy Calculation

NumPy helps us work with numbers easily.

Let's turn the `Final_Score` column into a NumPy array and calculate the **average**, **highest**, and **lowest** score.

```python
scores = np.array(df["Final_Score"])

print("Scores:", scores)

print("Average Score:", round(np.mean(scores), 1))
print("Highest Score:", np.max(scores))
print("Lowest Score:", np.min(scores))
```

## 📊 Simple Data Analysis

### 🏆 Who has the highest score?

Let's use Pandas to find the top student.

```python
top_student = df[df["Final_Score"] == df["Final_Score"].max()]
top_student
```

### 📚 Which students study more than 4 hours?

A simple example of filtering rows.

```python
hard_working = df[df["Study_Hours"] > 4]
hard_working[["Student", "Study_Hours", "Final_Score"]]
```

## 📈 Visualization 1 — Bar Chart

### 🏆 What is each student's final score?

Let's use a bar chart to compare all students' final scores at once.

```python
plt.figure(figsize=(9, 5))
plt.bar(df["Student"], df["Final_Score"], color="#4C72B0")
plt.title("Student Final Scores",c='g')
plt.xlabel("Student",c='r')
plt.ylabel("Final Score",c='r')
plt.show()
```

### 💡 What can we see?

We can quickly spot who scored the highest and who scored the lowest, just by looking at the bars.

## 📈 Visualization 2 — Line Chart

### 📚 Do more study hours mean better scores?

Let's use a line chart to compare study hours and final scores for each student.

```python
plt.figure(figsize=(9, 5))
plt.plot(df["Student"], df["Study_Hours"], marker="o", label="Study Hours")
plt.plot(df["Student"], df["Final_Score"] / 10, marker="o", label="Final Score (÷10)")
plt.title("Study Hours vs Final Score")
plt.xlabel("Student")
plt.ylabel("Value")
plt.legend(loc='lower right')
plt.show()
```

### 💡 What can we see?

The two lines tend to move together — students who study more hours generally have higher final scores. *(We divided the score by 10 just so both lines fit nicely on the same chart.)*

## 📈 Visualization 3 — Scatter Plot

### 🎯 Does attendance seem related to performance?

Let's use a scatter plot to see if attendance and final score are connected.

```python
plt.figure(figsize=(7, 5))
plt.scatter(df["Attendance"], df["Final_Score"], color="#0206BD")
plt.title("Attendance vs Final Score",c="g",fontweight="bold")
plt.xlabel("Attendance (%)",c="r")
plt.ylabel("Final Score",c="r")
plt.show()
```

### 💡 What can we see?

The dots go up from left to right. This means students with higher attendance generally have higher final scores.

## 📈 Visualization 4 — Simple Comparison Chart

### 📊 How does the class compare: Homework vs Previous Exam vs Final Exam?

Let's compare the **average** Homework Score, Previous Score, and Final Score using a bar chart.

```python
averages = [df["Homework_Score"].mean(), df["Previous_Score"].mean(), df["Final_Score"].mean()]
labels = ["Homework", "Previous Exam", "Final Exam"]

plt.figure(figsize=(7, 5))
plt.bar(labels, averages, color="#55A868", edgecolor="black")
plt.title("Average Score by Category",c="b")
plt.xlabel("Category",c="r")
plt.ylabel("Average Score",c="r")    
plt.show()
```

### 💡 What can we see?

This chart shows whether the class does better, worse, or about the same on homework, previous exams, and the final exam.
